# Feature Engineering

## Notebook Objective

Create analytical features that support revenue analysis, cohort analysis, churn analysis, and dashboard development.

## Features to Create

- Revenue
- Invoice Year
- Invoice Month
- Year-Month
- Invoice Day
- Invoice Hour

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
```

```python
# Load raw dataset
file_path = "../data/raw/online_retail_II.xlsx"

df = pd.read_excel(file_path)

# Remove duplicates
df = df.drop_duplicates(ignore_index=True)

# Remove accounting adjustment entries
df = df[df["Description"] != "Adjust bad debt"].reset_index(drop=True)

# Remove missing descriptions
df = df.dropna(subset=["Description"]).reset_index(drop=True)

print(f"Dataset ready for feature engineering: {df.shape}")
```

```python
# Business Feature:
# Revenue represents the monetary value of each transaction.
# Returns automatically become negative revenue because
# returned quantities are stored as negative values.

df["Revenue"] = df["Quantity"] * df["Price"]

print("Revenue column created successfully.")
```

```python
df[[
    "Quantity",
    "Price",
    "Revenue"
]].head(10)
```

```python
df[["Quantity", "Price", "Revenue"]].head(10)
```

```python
# Business Feature:
# Extract the year of each transaction for yearly trend analysis.

df["Invoice Year"] = df["InvoiceDate"].dt.year

print("Invoice Year feature created successfully.")
```

```python
df[["InvoiceDate", "Invoice Year"]].head(10)
```

```python
# Business Feature:
# Extract the transaction month for monthly trend analysis.

df["Invoice Month"] = df["InvoiceDate"].dt.month

print("Invoice Month feature created successfully.")
```

```python
df[["InvoiceDate", "Invoice Month"]].head(10)
```

```python
# Business Feature:
# Combine year and month into a single feature for monthly trend analysis.

df["Year-Month"] = df["InvoiceDate"].dt.to_period("M").astype(str)

print("Year-Month feature created successfully.")
```

```python
df[["InvoiceDate", "Invoice Year", "Invoice Month", "Year-Month"]].head(10)
```

```python
# Business Feature:
# Extract the day of the month for daily transaction analysis.

df["Invoice Day"] = df["InvoiceDate"].dt.day

print("Invoice Day feature created successfully.")
```

```python
df[["InvoiceDate", "Invoice Day"]].head(10)
```

```python
# Business Feature:
# Extract the transaction hour for hourly sales pattern analysis.

df["Invoice Hour"] = df["InvoiceDate"].dt.hour

print("Invoice Hour feature created successfully.")
```

```python
df[["InvoiceDate", "Invoice Hour"]].head(10)
```

```python
# Save the cleaned and feature-engineered dataset

output_path = "../data/cleaned/online_retail_cleaned.csv"

df.to_csv(output_path, index=False)

print(f"Dataset saved successfully to:\n{output_path}")
```
