```python
import pandas as pd
import numpy as np

# Display all columns
pd.set_option("display.max_columns", None)

# Display wider tables
pd.set_option("display.width", 1000)
```

```python
file_path = "../data/raw/online_retail_II.xlsx"

df = pd.read_excel(file_path)

df.head()
```

```python
# Display the number of rows and columns
print(f"Dataset Shape: {df.shape}")

print("\n")

# Display column names
print("Columns:")
print(df.columns.tolist())

print("\n")

# Display data types
print("Data Types:")
print(df.dtypes)
```

```python
df.info()
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

# Calculate percentage of missing values
missing_percentage = (missing_values / len(df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage.round(2)
})

# Display only columns with missing values
missing_summary[missing_summary["Missing Values"] > 0]
```

```python
# Check for duplicate rows
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Display duplicate rows (first 10)
df[df.duplicated()].head(10)
```

```python
# Count transactions with negative quantities
negative_quantity = df[df["Quantity"] < 0]

print(f"Negative quantity transactions: {len(negative_quantity)}")
```

```python
# View the first 10 negative quantity transactions
negative_quantity.head(10)
```

```python
# Business Check:
# Returns and cancellations are often recorded as negative quantities.
# We investigate them instead of removing them because they contribute
# to revenue leakage.

negative_quantity = df[df["Quantity"] < 0]
```

```python
# Business Check:
# Negative prices are unusual in retail transactions.
# We investigate them before deciding whether they are
# valid accounting entries or data quality issues.

negative_price = df[df["Price"] < 0]

print(f"Negative price transactions: {len(negative_price)}")
```
